```python
import numpy as np
np.random.seed(409) 
sales = np.random.randint(20, 80, size=(3, 4, 5))
stores = ['Downtown', 'Campus', 'Airport']
products = ['Latte', 'Espresso', 'Cold Brew', 'Sandwich', 'Cookie']
price = np.array([3.5, 3.0, 4.5, 6.5, 2.0]) 
```

```python
total_month = sales.sum()
print("1. 全月总销量:", total_month)

# 2. 分别沿单个轴求和
sum_axis0 = sales.sum(axis=0)  # 压缩「门店」维度 → 剩余(周, 产品)
sum_axis1 = sales.sum(axis=1)  # 压缩「周」维度 → 剩余(门店, 产品)
sum_axis2 = sales.sum(axis=2)  # 压缩「产品」维度 → 剩余(门店, 周)

print("\n2. 单轴求和结果形状:")
print(f"sum(axis=0) shape: {sum_axis0.shape} ")
print(f"sum(axis=1) shape: {sum_axis1.shape} ")
print(f"sum(axis=2) shape: {sum_axis2.shape} ")

# 3. 双轴聚合（一次性合并两个维度，本题核心考点）
total_per_store = sales.sum(axis=(1, 2))   # 每家门店总销量：合并 周+产品 两个轴
total_per_product = sales.sum(axis=(0, 1)) # 每种产品总销量：合并 门店+周 两个轴
total_per_week = sales.sum(axis=(0, 2))    # 每周总销量：合并 门店+产品 两个轴

print("\n3. 分类汇总结果:")
print(f"每家门店总销量: {total_per_store}")
print(f"每种产品总销量: {total_per_product}")
print(f"每周总销量: {total_per_week}")
```

我会和经理说：首先第4周和第2周只差3件，属于正常波动，而第1周销量高达844件，后面三周其实挤在一个很窄的区间里，所以我认为是"第1周特别好"，而不是"第4周特别差"。

```python
revenue = sales * price
print(revenue)
```

sales has shape (3, 4, 5)and price has shape (5,). Broadcasting right-aligns the shapes: price's single dimension (length 5) matches sales's last dimension (the product axis, also length 5). NumPy stretches price across the leading store and week axes, so each product's price is applied to that product's units in every store and every week. The 3.5 multiplies every Latte unit, 3.0 every Espresso, 4.5 every Cold Brew, 6.5 every Sandwich, and 2.0 every Cookie.

```python
revenue_per_store   = revenue.sum(axis=(1, 2))
revenue_per_product = revenue.sum(axis=(0, 1))

print("Revenue per store  :", revenue_per_store)
print("Revenue per product:", revenue_per_product)
```

按销量统计 Cookie 是最好的产品，它的总销量最高，受众最广、库存周转最快，是门店的核心流量产品。按营收统计 Sandwich 是最好的产品，它的单品单价更高，虽然销量不及 Cookie，但单份收入贡献更大，是门店的核心营收产品。两个排名出现差异的本质是单价差异，评估产品价值不能只看单一维度，需要结合销量、毛利、引流价值综合判断。

```python
cost = np.array([1.0, 1.2, 0.9])   # one number per store
sales * cost        
```

```python
# 修复
scaled = sales * cost.reshape(3, 1, 1)
print(scaled)
```

NumPy 广播遵循从右往左对齐维度的规则：price形状为(5,)，刚好和sales最右侧的产品维度（长度 5）长度一致，因此可以自动扩展维度完成相乘；而cost形状为(3,)，和sales最右侧维度（长度 5）长度不匹配且都不为 1，不符合广播规则因此报错。用reshape(3, 1, 1)调整后，cost的三个维度分别对应门店、周、产品，长度为 1 的维度会自动复制匹配，就能实现每家门店的所有销量数据乘以对应成本系数。

Checkpoint Questions
1、一共有 60 个数字。因为数组总元素数 = 各维度长度的乘积，即 3 × 4 × 5 = 60，对应 3 家门店、4 周、5 款产品，一共 60 个销售数据。
2、axis=1 代表「周」这个维度。比如对 axis=1 求和，就是把同一家门店、同一款产品的 4 周销量加总；简单说就是按周维度做聚合计算。
3、返回形状是 (3, 4)。结果里的单个数字代表：某一家门店、某一周，全部 5 款产品的总销量。
4、sales[1]：取第 1 号门店的全部数据，形状为 (4, 5)，也就是第 2 家店 4 周、5 款产品的所有销量。sales[:, 1]：取所有门店的第 1 周数据，形状为 (3, 5)，也就是 3 家店在第 2 周、全部产品的销量。二者一个是按「门店」切片，一个是按「周」切片。
5、因为 NumPy 广播是从右往左对齐维度：price 形状 (5,)，最右维度长度 5，和 sales 最右的产品维度长度 5 一致，符合广播规则。cost 形状 (3,)，最右维度长度 3，和 sales 最右维度长度 5 不匹配，且都不为 1，无法广播，因此报错。
6、我认为没有绝对的 “更好”，取决于业务目标：如果目标是引流、走量、提升门店客流，Cookie 更好；如果目标是提升营收和利润，Sandwich 更好。
会改变结论的因素包括：产品的毛利率、损耗率、人力制作成本、是否能带动其他产品的连带销售等。
7、以 sales.sum(axis=(1, 2)) 这行为例：sales：三维销量数组，存储 3 家门店、4 周、5 款产品的销售数量；.sum()：NumPy 的求和聚合函数，用于对数组内的数值加总；axis=(1, 2)：指定同时对第 1 轴（周）和第 2 轴（产品）求和，最终只保留门店维度，输出每家门店的全月总销量。
